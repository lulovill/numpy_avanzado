```python
import numpy as np
a = np.array([1, 2, 3, 4])

print (np.all(a != 0))
```

```python
# ejercicio 2

import numpy as np

a = np.array([0, 0, 2, 0])

print (np.any(a != 0))
```

```python
# ejercicio 3

import numpy as np

a = np.array([1, 2, np.nan, np.inf])

#valores = np.isnan(a)|np.isinf(a)

print (np.any( np.isnan(a)|np.isinf(a) ))
```

```python
# ejercicio 4

import numpy as np

a = np.array([1, 2, np.inf])

print (np.any(a = np.inf))
```

```python
# ejercicio 6

import numpy as np

a = np.array([1, 3, 5])
b = np.array([2, 3, 4])

# salio mal 
# print (a > b | a >= b | a < b | a <= b )

print(f"es a mayor que b:\n", 
      a > b)
print("es a mayoy/igual que b:\n", a >= b)
print("es a menor que b:\n",       a < b)
print("es a menor/igual que b:\n", a <= b)
```
